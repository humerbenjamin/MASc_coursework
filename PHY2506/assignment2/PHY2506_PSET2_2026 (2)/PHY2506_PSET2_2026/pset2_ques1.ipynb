```python
import numpy as np
import matplotlib.pyplot as plt

#Open file with the TES pressure levels and the retrieved and a prior profiles 
f1=open('TES_prof_sample1.dat','r')

#Read the first line with the lat/lon info and the number of retrieved levels
line = f1.readline().split()

# Get the number of retrieved levels, which is item 2
num_levs = int(line[1])

# Read the rest of the file into an array
data1 = np.fromfile(f1,dtype=float,sep=' ')
f1.close

# Extract the first num_levs values for the pressure levels
press_lev = data1[:num_levs]

# Extract the second num_levs values for the a priori profile
TES_prior = data1[num_levs:(num_levs+num_levs)]   

# Extract the last num_levs values for the retrieved profile
TES_prof = data1[(num_levs+num_levs):]

# Load the averaging kernel matrix
f2 = open('TES_avgkern_sample1.dat','r')
data2 = np.fromfile(f2,dtype=float,sep=' ')
TES_akern = np.reshape(data2,(num_levs,num_levs))
f2.close()

# Load the covariance matrix 
f3 = open('TES_covar_sample1.dat','r')
data3 = np.fromfile(f3,dtype=float,sep=' ')
TES_cov = np.reshape(data3,(num_levs,num_levs))
f3.close()

```
